```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = pd.read_csv('Student_Placement.csv')
```

```python
data.head()
```

```python
data.info()
```

```python
data['Gender'] = data['Gender'].map({'Male': 0, 'Female': 1})
```

```python
data['Placement_Offer'] = data['Placement_Offer'].map({'Yes': 1, 'No': 0})
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['Degree'] = le.fit_transform(data['Degree'])
```

```python
data.head()
```

```python
data.drop('Student_ID', axis=1, inplace=True)
```

```python
data.head()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Histogram for all columns
data.hist(figsize=(15, 12), bins=15, color='skyblue', edgecolor='black')

plt.suptitle('Distribution of All Variables', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Numerical columns for boxplots
numerical_cols = [
    'Age',
    'CGPA',
    'Technical_Skills_Score_100',
    'Communication_Skills_Score_100',
    'Aptitude_Test_Score_100',
    'Salary_Offered_USD'
]

# Create grid
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Flatten axes for easy looping
axes = axes.flatten()

# Loop through columns
for i, col in enumerate(numerical_cols):
    sns.boxplot(
        data=data,
        x='Placement_Offer',
        y=col,
        ax=axes[i]
    )

    axes[i].set_title(f'{col} vs Placement_Offer')

# Adjust layout
plt.tight_layout()

# Show all plots together
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

categorical_cols = [
    'Gender',
    'Degree',
    'Internships_Count',
    'Projects_Count',
    'Certifications_Count'
]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    sns.countplot(
        data=data,
        x=col,
        hue='Placement_Offer',
        ax=axes[i]
    )

    axes[i].set_title(f'{col} vs Placement_Offer')

# Remove empty subplot if any
fig.delaxes(axes[-1])

plt.tight_layout()
plt.show()
```
